```python
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Parameters for the Lissajous curve
A = 1  # Amplitude for x
B = 1  # Amplitude for y
a = 9  # Frequency for x
b = 8  # Frequency for y
delta = np.pi / 2  # Phase difference

T = 1
t = np.linspace(0, T*2*np.pi, 1000)

# Parametric equations
x = A * np.sin(a * t + delta)
y = B * np.sin(b * t)
```

```python
fig, ax = plt.subplots(facecolor='black', subplot_kw={'facecolor': 'black'})
ax.set_aspect('equal')
ax.grid(True, color='gray', linestyle='--', alpha=0.75)

ax.plot(x, y, color='limegreen', lw=2)
ax.set_xlim(-1.03, 1.03)
ax.set_ylim(-1.03, 1.03)

#ax.axis('off')
ax.set_xticklabels([])
ax.set_yticklabels([])

plt.show()
```
